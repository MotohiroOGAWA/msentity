# Basic usage of msentity

Load, inspect, filter, and write mass-spectrometry datasets.

## Load and inspect a dataset

`load_ms_dataset` detects MSP, MGF, MSDS, TSV, and CSV from the filename.

```
import pandas as pd
from msentity import load_ms_dataset

dataset = load_ms_dataset("example.msp")
dataset.metadata.head()
```

## Work with records and peaks

```
record = dataset[0]
print(record["Name"], record["PrecursorMZ"], record.n_peaks)
print(record.spectrum.mz)
print(record.spectrum.intensity)
```

## Filter and sort

```
precursor_mz = pd.to_numeric(dataset["PrecursorMZ"], errors="coerce")
filtered = dataset[precursor_mz > 300]
filtered = filtered.sort_by("PrecursorMZ", ascending=False)
filtered.metadata
```

## Save datasets

MSDS preserves the complete dataset. MSP, MGF, TSV, and CSV are interchange formats.

```
from pathlib import Path
import tempfile
from msentity import write_csv, write_mgf, write_msp, write_tsv

with tempfile.TemporaryDirectory() as directory:
    output = Path(directory)
    filtered.save(output / "filtered.msds")
    write_msp(filtered, output / "filtered.msp")
    write_mgf(filtered, output / "filtered.mgf")
    write_tsv(filtered, output / "filtered.tsv")
    write_csv(filtered, output / "filtered.csv")
```
